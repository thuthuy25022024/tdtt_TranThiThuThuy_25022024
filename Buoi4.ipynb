```python
# W4A7

import math

def is_prime(num):
    if num < 2:
        return False
    for i in range(2, int(math.sqrt(num)) + 1):
        if num % i == 0:
            return False
    return True

while True:
    n = int(input("Nhập số nguyên dương n (n ≥ 2): "))
    if n >= 2:
        break
    print("Vui lòng nhập n ≥ 2.")

# Tìm ước nguyên tố lớn nhất
for i in range(n, 1, -1):
    if n % i == 0 and is_prime(i):
        print("Ước nguyên tố lớn nhất của", n, "là:", i)
        break
```

```python
# W4A8

def is_palindrome(n):
    return str(n) == str(n)[::-1]  # kiểm tra số đối xứng

def reverse_number(n):
    return int(str(n)[::-1])  # đảo ngược số

def tinh_palindrome(n):
    buoc = 0
    while not is_palindrome(n):  # khi nào chưa là số đối xứng
        n = n + reverse_number(n)  # cộng với số đảo ngược
        buoc += 1
    return buoc, n

n = int(input("Nhập số nguyên dương n: "))
if n <= 0:
    print("Vui lòng nhập số nguyên dương > 0.")
else:
    buoc, so_palindrome = tinh_palindrome(n)
    print("Số bước:", buoc)
    print("Số palindrome thu được:", so_palindrome)
```

```python
# W4A9

import math

def cac_chu_so_khac_nhau(num):
    chu_so = str(num)
    return len(set(chu_so)) == len(chu_so)

def in_so_chinh_phuong(n):
    co_so = False
    can = int(math.sqrt(n))

    for i in range(1, can + 1):
        bp = i * i
        if bp <= n and cac_chu_so_khac_nhau(bp):
            print(bp, end=' ')
            co_so = True

    if not co_so:
        print("(no number)")

n = int(input("Nhập số nguyên dương n: "))
if n <= 0:
    print("Vui lòng nhập số nguyên dương.")
else:
    in_so_chinh_phuong(n)
```

```python
# W4A10

def collatz_length(x):
    count = 1
    while x != 1:
        if x % 2 == 0:
            x //= 2
        else:
            x = 3 * x + 1
        count += 1
    return count

n = int(input("Nhập số nguyên dương n: "))
if n <= 0:
    print("Vui lòng nhập số nguyên dương.")
else:
    max_len = 0
    so_nho_nhat = 1

    for i in range(1, n + 1):
        length = collatz_length(i)
        if length > max_len:
            max_len = length
            so_nho_nhat = i
        elif length == max_len and i < so_nho_nhat:
            so_nho_nhat = i

    print(so_nho_nhat, max_len)
```

```python
# W4A11

import math

def dem_uoc_so_chan(n):
    count = 0
    can = int(math.sqrt(n))

    for i in range(1, can + 1):
        if n % i == 0:
            # i là ước số
            if i % 2 == 0:
                count += 1
            # n//i là ước số khác i
            other = n // i
            if other != i and other % 2 == 0:
                count += 1
    return count

n = int(input("Nhập số nguyên dương n (<10^6): "))
if n <= 0 or n >= 10**6:
    print("Vui lòng nhập số nguyên dương nhỏ hơn 10^6.")
else:
    print(dem_uoc_so_chan(n))
```

```python
# W4A12

def tinh_tien_lai(X, N):
    r = 0.007  # lãi suất tháng
    so_tien = X * (1 + r) ** N
    return int(so_tien)  # bỏ phần thập phân

X = int(input("Nhập số tiền gửi ban đầu X: "))
N = int(input("Nhập số tháng gửi N: "))

if X < 0 or N < 0:
    print("Vui lòng nhập số dương.")
else:
    ket_qua = tinh_tien_lai(X, N)
    print(f"Số tiền sau {N} tháng là: {ket_qua} đồng")
```

```python
# W4A13

def tong_uoc(n):
    total = 1  # 1 luôn là ước số (ngoại trừ số 1 thì tổng ước là 0)
    if n == 1:
        return 0  # số 1 không có ước khác ngoài chính nó
    can = int(n**0.5)
    for i in range(2, can + 1):
        if n % i == 0:
            total += i
            if i != n // i:
                total += n // i
    return total

a = int(input("Nhập số thứ nhất: "))
b = int(input("Nhập số thứ hai: "))

if a <= 0 or b <= 0:
    print("Vui lòng nhập số nguyên dương.")
else:
    if tong_uoc(a) == b and tong_uoc(b) == a:
        print("true")
    else:
        print("false")
```

```python
# W4A15

def tim_ga_cho(total_animals, total_legs):
    # Tính số chó y
    y = (total_legs - 2 * total_animals) / 2
    # Tính số gà x
    x = total_animals - y

    if y.is_integer() and y >= 0 and x >= 0:
        print(f"Gà: {int(x)}, Chó: {int(y)}")
    else:
        print("invalid")

total_animals = int(input("Nhập tổng số con: "))
total_legs = int(input("Nhập tổng số chân: "))

tim_ga_cho(total_animals, total_legs)
```

```python
# W4A23

n = int(input("Nhập số nguyên dương n: "))

if n <= 1:
    print("Không có k nào thỏa điều kiện.")
else:
    k = 0      # Bắt đầu từ 0
    S = 0      # Tổng ban đầu là 0

    while True:
        k += 1       # Tăng k lên 1
        S += k       # Cộng k vào tổng S

        if S >= n:   # Nếu tổng đã lớn hơn hoặc bằng n thì dừng
            k -= 1   # Giảm k về giá trị trước đó
            break

    print("Giá trị k lớn nhất sao cho S(k) < n là:", k)
```

```python
# W4A27

s = input("Nhập chuỗi: ")

# Tính độ dài chuỗi bằng cách duyệt từng ký tự
length = 0
for _ in s:
    length += 1

count = 0
for i in range(length):
    # Kiểm tra ký tự hiện tại là chữ cái
    if (('a' <= s[i] <= 'z') or ('A' <= s[i] <= 'Z')):
        # Nếu là ký tự đầu tiên hoặc ký tự trước là khoảng trắng thì tăng đếm
        if i == 0 or s[i-1] == ' ':
            count += 1

print("Số từ trong chuỗi là:", count)
```

```python
# W4A28

s = input("Nhập chuỗi: ")

# Tính độ dài chuỗi thủ công (không dùng len)
length = 0
for _ in s:
    length += 1

word = ""
found_word = False

# Duyệt chuỗi theo index từ 0 đến length - 1
for i in range(length):
    c = s[i]
    if not found_word:
        if c != ' ':
            found_word = True
            word += c
    else:
        if c == ' ':
            break
        else:
            word += c

print("Từ đầu tiên trong chuỗi là:", word)
```

```python
# W4A32

s = input("Nhập mật khẩu: ")

has_upper = False
has_lower = False
has_digit = False
has_special = False

special_chars = "!@#$%^&*()-_=+[]{}|;:'\",.<>/?`~"

for c in s:
    if 'A' <= c <= 'Z':
        has_upper = True
    elif 'a' <= c <= 'z':
        has_lower = True
    elif '0' <= c <= '9':
        has_digit = True
    elif c in special_chars:
        has_special = True

if len(s) > 6 and has_upper and has_lower and has_digit and has_special:
    print("Đây là một mật khẩu mạnh.")
else:
    print("Đây KHÔNG phải là một mật khẩu mạnh.")
```

```python
# W4A33

n = int(input("Nhập số nguyên: "))

s = str(n)  # Chuyển số sang chuỗi
result = ""

count = 0
for i in range(len(s) - 1, -1, -1):  # Duyệt ngược từ cuối chuỗi về đầu
    result = s[i] + result
    count += 1
    if count % 3 == 0 and i != 0:
        result = '.' + result

print("Đổi sang chuỗi rồi in ra:", result)
```
