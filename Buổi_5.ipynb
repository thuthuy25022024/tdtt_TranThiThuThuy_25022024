```python
# W5A3
n = int(input())
def so_nguyen_to(n):
  if n < 2:
    return False
  else:
    for i in range(2, int(n**0.5) + 1):
      if n % i == 0:
        return False
    return True
print(so_nguyen_to(n))
```

```python
# W5A4
n = int(input())
def so_hoan_hao(n):
  tong = 0
  for i in range(1, n):
    if n % i == 0:
      tong += i
  return tong == n
print(so_hoan_hao(n))
```

```python
# W5A5
s = list(map(int, input().split()))
k = int(input())
def vi_tri_dau_tien(s, k):
  for i in range(len(s)):
    if s[i] == k:
      return i
  return -1
print(vi_tri_dau_tien(s, k))
```

```python
# W5A6
n = int(input())
def giai_thua(n):
  gt = 1
  for i in range (1, n + 1):
    gt *= i
  return gt
print(giai_thua(n))
```

```python
# W5A7
num1 = float(input())
operator = input()
num2 = float(input())

def may_tinh_bo_tui(num1, num2, operator):
    if operator == '+':
        kq = num1 + num2
    elif operator == '-':
        kq = num1 - num2
    elif operator == '*':
        kq = num1 * num2
    elif operator == '/':
        if num2 == 0:
            return 'Lỗi'
        else:
          kq = num1 / num2
    else:
        return 'Lỗi'
    return f'{kq:.2f}'
print(may_tinh_bo_tui(num1, num2, operator))
```

```python
# W5A8
def kc_hamming(x, y):
  return bin(x ^ y).count('1')
x, y = map(int, input().split())
print(kc_hamming(x, y))
```

```python
# W5A9
n = int(input())
def tong(n):
  tong = 0
  while n > 0:
    tong += n % 10
    n = n // 10
  return tong
print(tong(n))
```

```python
# W5A10
a, b = input().split()
def dang_cau(a, b):
  mapping = {}
  used_values = set()
  for x, y in zip(a, b):
    if x in mapping:
      if mapping[x] != y:
        return False
    else:
      if y in used_values:
        return False
      mapping[x] = y
      used_values.add(y)
  return True
print(dang_cau(a, b))
```
