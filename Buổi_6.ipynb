```python
# W6A4
s = input().split()
kq = {}
for pair in s:
  key, value = pair.split(':')
  if key in kq:
    kq[key].append(value)
  else:
    kq[key] = [value]
print(kq)
```

```python
# W6A9
d1 = {}
data1 = input().split()
for pair in data1:
  key, value = pair.split(':')
  d1[key] = int(value)

d2 = {}
data2 = input().split()
for pair in data2:
  key, value = pair.split(':')
  d2[key] = int(value)

d3 = {}

for key, value in d1.items():
  if key in d2:
    d3[key] = value + d2[key]
  else:
    d3[key] = value

for key, value in d2.items():
  if key not in d3:
    d3[key] = value

d3 = dict(sorted(d3.items()))
print(d3)
```

```python
# W6A10
s = list(map(int, input().split()))
k = int(input())
kq = []
for i in range(len(s)):
  for j in range(i + 1, len(s)):
    if s[i] + s[j] == k:
      kq.append((s[i],s[j]))
kq.sort(key = lambda x: x[0])
print(kq)
```

```python
# W6A12
s = list(map(int, input().split()))
dem = {}
for i in s:
  if i in dem:
    dem[i] += 1
  else:
    dem[i] = 1
max_dem = max(dem.values())
kq = [key for key, value in dem.items() if value == max_dem]
print(*kq)

```

```python
# W6A15
d = {}
data = input().split()
for i in range(0, len(data), 2):
  key = data[i]
  value = int(data[i + 1])
  d[key] = value
k= int(input())
kq = {}
for key, value in d.items():
  if value > k:
    kq[key] = value
print(kq)
```

```python
# W6A16
# Nhập n, m
n, m = map(int, input().split())

# Nhập mảng
a = []
for i in range(n):
    row = list(map(int, input().split()))
    a.append(row)

# In ma trận với mỗi phần tử rộng 4, căn phải
for i in range(n):
    for j in range(m):
        print(f"{a[i][j]:>4}", end="")
    print()
```

```python
# W6A17
# Nhập n
n = int(input())

# Nhập mảng vuông n x n
a = []
for i in range(n):
    row = list(map(int, input().split()))
    a.append(row)

# In đường chéo chính
print("Duong cheo chinh:")
for i in range(n):
    print(a[i][i], end=" ")
print()

# In đường chéo phụ
print("Duong cheo phu:")
for i in range(n):
    print(a[i][n - 1 - i], end=" ")
```

```python
# W6A18
n, m, k = map(int, input().split())
a = []
for i in range(n):
    row = list(map(int, input().split()))
    a.append(row)

tong_cot = 0
for i in range(n):
    tong_cot += a[i][k - 1]
tong_hang = 0
for i in range(m):
    tong_hang += a[k - 1][i]
print(tong_cot)
print(tong_hang)
```
